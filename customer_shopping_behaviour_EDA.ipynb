**Checking the null values for all the columns as we can see here Review Rating columns has some null values present in it**

```python
#checking the null values
df.isnull().sum()
# so here we can see that review rating column has null values
```

**Here Iam filling the null values via using median of the review rating column for more accuracy avoiding filling it randomly**

```python
# here what i did withe review rating is that replacing the null values by the median values rather just filling the nulls
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x : x.fillna(x.median()))
```

### Displaying and checking the nulls are filled

```python
# checking the code is worked or not for the nulls
df.isnull().sum()
```

### Renaming the column name

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ","_")
df = df.rename(columns={'purchase_amount_(usd)':'purchase_usd'})
```

```python
df.columns
```

### Creating the labels for the age group column

```python
labels = ['young_adult','prime','middle_aged','senior']
df['age_group']= pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age','age_group']].head()
```

### Doing the frequency mapping for the customer buying behaviour

```python
frequency_mapping = {
    'fortnightly': 14,
    'weekly': 7,
    'bi-weekly': 14,
    'monthly': 30,
    'quarterly': 90,
    'every 3 months': 90,
    'annually': 365,
}

df['purchase_frequency_days'] = (
    df['frequency_of_purchases']
    .str.strip()
    .str.lower()
    .map(frequency_mapping)
)
```
